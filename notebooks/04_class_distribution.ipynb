# 04 — Class Distribution & Negative Sampling Evaluation

Examines class imbalance ratios across train, validation, and test splits.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_theme(style="whitegrid")

train = pd.read_parquet("../data/processed/train.parquet")
val = pd.read_parquet("../data/processed/validation.parquet")
test = pd.read_parquet("../data/processed/test.parquet")

splits_summary = pd.DataFrame([
    {"Split": "Train (2016–2022)", "Total Rows": len(train), "Positives": train['fire_occurred'].sum(), "Positive %": train['fire_occurred'].mean()*100},
    {"Split": "Validation (2023–2024)", "Total Rows": len(val), "Positives": val['fire_occurred'].sum(), "Positive %": val['fire_occurred'].mean()*100},
    {"Split": "Test (2025 Holdout)", "Total Rows": len(test), "Positives": test['fire_occurred'].sum(), "Positive %": test['fire_occurred'].mean()*100},
])
print(splits_summary.to_string(index=False))
```

## 1. Class Distribution Bar Chart


```python
fig, ax = plt.subplots(figsize=(9, 5))
sns.barplot(data=splits_summary, x="Split", y="Positive %", palette="magma", ax=ax)
ax.set_title("Positive Fire Label Ratio Across Splitting Phases", fontsize=13, fontweight="bold")
ax.set_ylabel("Positive Label Percentage (%)")
for p in ax.patches:
    ax.annotate(f"{p.get_height():.2f}%", (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='bottom', fontsize=10, xytext=(0, 3), textcoords='offset points')
plt.tight_layout()
plt.show()
```

## 2. Key Takeaways


- **Train Split (20:1 Undersampled)**: Positives represent ~4.76% of training observations, optimizing training speed while retaining all positive instances.
- **Validation & Test (Full Natural Distribution)**: Positives represent ~0.8% - 1.2% of municipality-days, ensuring model evaluations reflect real-world operational prevalence.
