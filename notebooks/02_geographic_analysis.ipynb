# 02 — Geographic & Bioclimatic Analysis of Wildfires

This notebook explores spatial clustering, state/municipality breakdown, and bioclimatic zone distribution.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_theme(style="whitegrid")
DATA_PATH = Path("../data/interim/fires_harmonized.parquet")
df = pd.read_parquet(DATA_PATH)
```

## 1. Top States by Fire Volume


```python
state_counts = df['estado'].value_counts().head(15)

fig, ax = plt.subplots(figsize=(10, 6))
sns.barplot(x=state_counts.values, y=state_counts.index, ax=ax, palette="flare")
ax.set_title("Top 15 States by Total Wildfire Incidents (2015–2025)", fontsize=14, fontweight="bold")
ax.set_xlabel("Number of Fires")
for p in ax.patches:
    ax.annotate(f"{int(p.get_width()):,}", (p.get_width(), p.get_y() + p.get_height()/2.),
                ha='left', va='center', fontsize=10, xytext=(5, 0), textcoords='offset points')
plt.tight_layout()
plt.show()
```

## 2. Top Municipalities


```python
muni_counts = df.groupby(['estado', 'municipio']).size().sort_values(ascending=False).head(15)
muni_labels = [f"{m} ({e})" for e, m in muni_counts.index]

fig, ax = plt.subplots(figsize=(10, 6))
sns.barplot(x=muni_counts.values, y=muni_labels, ax=ax, palette="viridis")
ax.set_title("Top 15 Municipalities by Wildfire Incidents", fontsize=14, fontweight="bold")
ax.set_xlabel("Number of Fires")
plt.tight_layout()
plt.show()
```

## 3. Geographic Coordinate Distribution


```python
fig, ax = plt.subplots(figsize=(9, 6))
scatter = ax.scatter(df['longitud'], df['latitud'], c=df['fecha_inicio'].dt.year,
                     cmap="plasma", s=2, alpha=0.4)
ax.set_title("Geographic Distribution of Wildfire Centroids", fontsize=14, fontweight="bold")
ax.set_xlabel("Longitude (°W)")
ax.set_ylabel("Latitude (°N)")
plt.colorbar(scatter, label="Year")
plt.tight_layout()
plt.show()
```

## 4. Key Findings Summary


- **Regional Hotspots**: States such as México, Michoacán, Jalisco, Mexico City, and Chihuahua account for the highest concentration of fire events.
- **Concentrated Spatial Risk**: Top municipalities exhibit recurring fire risk due to mountain/forest topographies combined with high population interface.
