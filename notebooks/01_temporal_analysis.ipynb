# 01 — Temporal Pattern Analysis of Wildfires in Mexico (2015–2025)

This notebook analyzes temporal patterns in the harmonized wildfire dataset:
- Annual trends and fire counts
- Monthly seasonality (peak dry season)
- Day-of-week distributions
- Fire duration dynamics


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams["figure.figsize"] = (10, 5)

DATA_PATH = Path("../data/interim/fires_harmonized.parquet")
df = pd.read_parquet(DATA_PATH)
print(f"Total harmonized fire records: {len(df):,}")
print(f"Date range: {df['fecha_inicio'].min()} to {df['fecha_inicio'].max()}")
```

## 1. Annual Fire Counts


```python
yearly = df.groupby("anio").size()

fig, ax = plt.subplots(figsize=(10, 5))
sns.barplot(x=yearly.index, y=yearly.values, ax=ax, palette="Blues_d")
ax.set_title("Wildfire Events per Year in Mexico (2015–2025)", fontsize=14, fontweight="bold")
ax.set_xlabel("Year")
ax.set_ylabel("Number of Recorded Fires")
for p in ax.patches:
    ax.annotate(f"{int(p.get_height()):,}", (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='bottom', fontsize=10, xytext=(0, 3), textcoords='offset points')
plt.tight_layout()
plt.show()
```

## 2. Monthly Seasonality


```python
df['month'] = df['fecha_inicio'].dt.month
monthly = df.groupby('month').size()
month_names = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']

fig, ax = plt.subplots(figsize=(10, 5))
sns.barplot(x=month_names, y=monthly.values, ax=ax, color="#e74c3c")
ax.set_title("Wildfire Seasonality: Monthly Fire Counts (2015–2025)", fontsize=14, fontweight="bold")
ax.set_xlabel("Month")
ax.set_ylabel("Total Recorded Fires")
pct_jan_jun = (df['month'].isin(range(1, 7)).sum() / len(df)) * 100
ax.text(0.02, 0.88, f"Jan–Jun accounts for {pct_jan_jun:.1f}% of all fires", transform=ax.transAxes,
        fontsize=11, bbox=dict(boxstyle="round,pad=0.3", facecolor="white", alpha=0.8))
plt.tight_layout()
plt.show()
```

## 3. Day of Week Patterns


```python
df['day_of_week'] = df['fecha_inicio'].dt.day_name()
days_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
dow_counts = df['day_of_week'].value_counts().reindex(days_order)

fig, ax = plt.subplots(figsize=(9, 4))
sns.barplot(x=dow_counts.index, y=dow_counts.values, ax=ax, color="#3498db")
ax.set_title("Wildfire Distribution by Day of Week", fontsize=14, fontweight="bold")
ax.set_ylabel("Total Fires")
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

## 4. Key Findings Summary


- **Extreme Seasonality**: Over 90% of wildfire ignitions occur between January and June, peaking in April and May during the dry season.
- **Stable Multi-Year Baseline**: Annual fire counts range between 3,800 and 8,900 incidents per year.
- **Human Factor**: Day-of-week distributions show mild weekend increases, supporting historical findings that ~98% of Mexican forest fires have human-related causes.
